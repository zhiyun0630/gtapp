```python
import streamlit as st
import time

# 1. 页面基础配置（让页面看起来更专业）
st.set_page_config(page_title="地质AI辅助解译系统", page_icon="🪨", layout="wide")

# 2. 侧边栏 (Sidebar) - 用于放置系统设置
st.sidebar.title("⚙️ 系统设置")
model_choice = st.sidebar.selectbox("选择AI大模型", ["通义千问-VL", "智谱 GLM-4V", "Kimi"])
standard_choice = st.sidebar.radio("遵循地质规范", ["DZ/T0001-91 (区域地质调查)", "DZ/T0261 (矿山地质)"])
st.sidebar.info(f"当前系统版本：V1.0.0 \n开发人员：[你的名字]")

# 3. 主界面标题
st.title("🪨 地质AI辅助解译与成图系统")
st.markdown("---") # 画一条分割线

# 4. 核心功能区：分两列显示
col1, col2 = st.columns(2)

with col1:
    st.subheader("📥 1. 数据输入")
    # 上传文件组件
    uploaded_file = st.file_uploader("请上传野外露头或无人机照片", type=['jpg', 'png', 'jpeg'])
    
    if uploaded_file is not None:
        # 显示上传的图片
        st.image(uploaded_file, caption="已上传图像", use_column_width=True)

with col2:
    st.subheader("🤖 2. AI 智能解译")
    if uploaded_file is not None:
        # 模拟 AI 思考的过程
        with st.spinner("AI 正在结合规范进行地质特征提取，请稍候..."):
            time.sleep(2) # 假装 AI 在计算，停顿2秒
            
        # 模拟 AI 生成的结果
        ai_result = """**【AI初步解译结果】**
*   **岩性定名**：中细粒黑云母二长花岗岩
*   **颜色**：浅肉红色
*   **结构构造**：中细粒花岗结构，块状构造
*   **矿物成分**：钾长石(45%)、斜长石(25%)、石英(20%)、黑云母(10%)
*   **风化特征**：表面见轻微球形风化。
"""
        st.success("解译完成！")
        
        # 核心亮点：人工干预组件（允许地质同学修改AI的幻觉）
        st.markdown("👨‍🔬 **3. 专家人工审核与修正**")
        final_result = st.text_area("请核对上方结果，如有误请在此修改：", value=ai_result, height=250)

# 5. 底部操作区
st.markdown("---")
if st.button("📥 导出标准化地质报告 (Word)", type="primary", use_container_width=True):
    st.balloons() # 点击后放气球庆祝
    st.info("报告生成成功！已保存至本地目录 (此处后续接入 python-docx 代码)")
```
